```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
import pandas as pd

url = "https://raw.githubusercontent.com/Irfansafi2026/advanced-math-project/main/GASDESM%20.csv"

df = pd.read_csv(url)
df.head()
```

```python
df['observation_date'] = pd.to_datetime(df['observation_date'])

df = df[df['observation_date'].dt.year.between(2020, 2025)].copy()

df = df.dropna(subset=['GASDESM'])

print(df.shape)
df.head()
```

```python
print("Average Price:", df['GASDESM'].mean())
print("Maximum Price:", df['GASDESM'].max())
print("Minimum Price:", df['GASDESM'].min())
print("Standard Deviation:", df['GASDESM'].std())
```

```python
monthly_miles = 10000
mpg = 6

df['Fuel_Cost'] = (monthly_miles / mpg) * df['GASDESM']

print("Average Monthly Cost:", df['Fuel_Cost'].mean())
print("Maximum Monthly Cost:", df['Fuel_Cost'].max())
print("Minimum Monthly Cost:", df['Fuel_Cost'].min())
```

```python
plt.figure(figsize=(10, 5))
plt.plot(df['observation_date'], df['GASDESM'])
plt.title('US Diesel Prices (2020-2025)')
plt.xlabel('Year')
plt.ylabel('Price per Gallon ($)')
plt.grid(True)
plt.tight_layout()
plt.savefig('diesel_prices.pdf')
plt.show()
```

```python
plt.figure(figsize=(10, 5))
plt.bar(df['observation_date'].dt.year,
        df['Fuel_Cost'].groupby(df['observation_date'].dt.year).transform('mean'))
plt.title('Average Monthly Truck Fuel Costs by Year')
plt.xlabel('Year')
plt.ylabel('Fuel Cost ($)')
plt.tight_layout()
plt.savefig('truck_fuel_costs.pdf')
plt.show()
```
